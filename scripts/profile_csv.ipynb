```python


# 1. SRID DATA PROFILING

def profile_srid(file_path):

    df = pd.read_csv(file_path)

    print("\nSRID PROFILE")

    # Check the date range
    dates = pd.to_datetime(df["date"], errors="coerce")
    print("\nDATE RANGE:", dates.min(), "to", dates.max())

    # Check the distribution of important categories
    print("\nTYPE:", df["Type"].value_counts().to_dict())
    print("\nSOURCE:", df["source"].value_counts().head(10).to_dict())
    print("\nCOMMODITIES:", df["commodity"].value_counts().to_dict())

    # Check the number of unique markets and regions
    print("\nMARKETS:", df["Market"].nunique())
    print("\nREGIONS:", df["region"].unique().tolist())

    # Inspect geolocation information
    print("\nGEOLOCATION SAMPLE:", df["_geolocation"].head().tolist())
    print("GEOLOCATION UNIQUE COUNT:", df["_geolocation"].nunique())

    # Identify missing values
    print("\nNULL COUNTS:\n", df.isnull().sum())

    # Check duplicate records using the proposed combination of columns
    duplicates = df.duplicated(
        subset=["Market", "commodity", "date", "Type"]
    ).sum()

    print("\nDUPLICATES:", duplicates)



# 2. WFP DATA PROFILING


def profile_wfp(file_path):

    df = pd.read_csv(file_path)

    print("\n WFP PROFILE ")

    # Check the date range
    dates = pd.to_datetime(df["date"], errors="coerce")
    print("\nDATE RANGE:", dates.min(), "to", dates.max())

    # Examine currency, units and price classifications
    print("\nCURRENCY:", df["currency"].value_counts().to_dict())
    print("\nUNIT:", df["unit"].value_counts().to_dict())
    print("\nPRICE TYPE:", df["pricetype"].value_counts().to_dict())
    print("\nPRICE FLAG:", df["priceflag"].value_counts().to_dict())

    # Examine categories and commodities
    print("\nCATEGORY:", df["category"].value_counts().to_dict())
    print("\nCOMMODITIES:", df["commodity"].value_counts().head(15).to_dict())

    # Check the number of unique markets
    print("\nMARKETS:", df["market"].nunique())

    # Identify missing values
    print("\nNULL COUNTS:\n", df.isnull().sum())



# 3. FAOSTAT DATA PROFILING


def profile_fao(file_path):

    df = pd.read_csv(file_path)

    print("\n FAOSTAT PROFILE ")

    # Check the range of available years
    print("\nYEAR RANGE:", df["Year"].min(), "to", df["Year"].max())

    # Examine items, elements and measurement units
    print("\nITEMS:", df["Item"].value_counts().to_dict())
    print("\nELEMENTS:", df["Element"].value_counts().to_dict())
    print("\nUNITS:", df["Unit"].value_counts().to_dict())

    # Examine data quality flags
    print("\nFLAGS:", df["Flag"].value_counts().to_dict())
    print("\nFLAG DESCRIPTIONS:", df["Flag Description"].value_counts().to_dict())

    # Check the available domains
    print("\nDOMAINS:", df["Domain"].unique().tolist())

    # Identify missing values
    print("\nNULL COUNTS:\n", df.isnull().sum())




profile_srid(srid)

```

```python


# Load the SRID dataset
df = pd.read_csv(srid)


# ==========================================
# 1. MARKETS AVAILABLE IN EACH REGION
# ==========================================

print("\n========== MARKETS BY REGION ==========")

# Get all regions, excluding missing values
regions = sorted(df["region"].dropna().unique())

for region in regions:

    # Get unique markets belonging to the current region
    markets = df[df["region"] == region]["Market"].dropna().unique()

    print(f"\n{region} ({len(markets)} markets):")
    print(list(markets)[:15])


# ==========================================
# 2. SEARCH FOR MAJOR NORTHERN CITIES
# ==========================================

print("\n========== NORTHERN CITIES SEARCH ==========")

# Cities we want to check for
cities = [
    "Tamale", "Yendi", "Savelugu",
    "Bolgatanga", "Wa", "Nalerigu", "Damongo"
]

for city in cities:

    # Find market names containing the city name
    matches = df[
        df["Market"].astype(str).str.contains(city, case=False, na=False)
    ]

    # Display the regions where matching markets were found
    if not matches.empty:
        print(f"\n{city}: Found in {matches['region'].unique().tolist()}")
    else:
        print(f"\n{city}: Not found")
```

```python


# Load both datasets
srid = pd.read_csv(r'C:\Users\DELL\Projects\agrolens\data\Commodity prices _04.11.25.csv')
wfp = pd.read_csv(r'C:\Users\DELL\Projects\agrolens\data\wfp_food_prices_gha.csv')


# ==========================================
# 1. EXTRACT AND STANDARDISE MARKET NAMES
# ==========================================

# Remove missing values, extra spaces and differences in letter casing
srid_markets = set(
    srid["Market"].dropna().str.strip().str.lower()
)

wfp_markets = set(
    wfp["market"].dropna().str.strip().str.lower()
)


# ==========================================
# 2. COMPARE THE TWO MARKET LISTS
# ==========================================

# Find markets appearing in both datasets
overlap = srid_markets & wfp_markets

# Find markets appearing in only one dataset
srid_only = srid_markets - wfp_markets
wfp_only = wfp_markets - srid_markets


# ==========================================
# 3. DISPLAY RESULTS
# ==========================================

percentage = (len(overlap) / len(srid_markets)) * 100 if srid_markets else 0

print("\n========== MARKET COMPARISON ==========")

print(f"\nSRID markets: {len(srid_markets)}")
print(f"WFP markets: {len(wfp_markets)}")

print(f"\nOverlapping markets: {len(overlap)}")
print(f"SRID overlap percentage: {percentage:.0f}%")

print("\nMarkets appearing in both datasets:")
print(sorted(overlap))

print("\nMarkets only in SRID (first 20):")
print(sorted(srid_only)[:20])

print("\nMarkets only in WFP (first 20):")
print(sorted(wfp_only)[:20])
```
